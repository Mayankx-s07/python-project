```python
  #                                         ***Automated Tech Campus Multi-Tier Billing and Pass Engine***

print("--------------------------------------------------------")
print("  Module A: User Classification & Base Pass Tariff")
print("--------------------------------------------------------")

a=int(input("Select category (1:student ,2:faculty/staff ):"))
basefee = 0
discount = 0
discount_percent = 0
p_fee = 0
peak_surcharge = 0
fixed_charge = 0
e_bill = 0
#____________________for students________________________
if a==1:
    print("Student selected")
    b=input("Enter sub category (UG/PG):").upper()
    if b=="UG":
        basefee=500
    elif b=="PG":
        basefee=350
    else:
        print("Invalid sub category")
#____________________for faculty/staff___________________
elif a==2:
    print("faculty/staff selected")
    b=int(input(" Enter sub category (1:visiting ,2:Resident):"))
    if b==2:
        basefee=800
    elif b==1:
        basefee=1200
    else:
        print("Invalid sub category")
else:
    print("Invalid category")
print("--------------------------------------------------------")
print("        Module B: Surcharge & Discount Rules ")
print("--------------------------------------------------------")

#____________________for students________________________
if a==1:
    cgpa=float(input("Enter Student CGPA (0.0 - 10.0): "))
    if cgpa<0 or cgpa>10.00:
        print("error , Invalid cgpa")
        exit()
    elif cgpa>=8.5:
        discount=0.20*basefee
    elif cgpa>=7.5:
        discount=0.10*basefee
    else:
        discount=0
#____________________for faculty/staff___________________
elif a==2:
    years_service=int(input("Enter years of service"))
    if years_service<0:
        print("service year cant be negative")
        exit()
    elif years_service>10:
        discount=0.15*basefee
    else:
        discount=0
# net pass fee
net_pass_fee=(basefee-discount)

#____________________for Parking permit____________________
p=int(input("Select parking permit""(0:none,2:two wheeler,4:four wheeler):"))

if p==0:
    p_fee=0
elif p==2:
    p_fee=200
elif p==4:
    p_fee=600
else:
    print("Invalid parking option")

# Student with 4 wheeeler peak-hour surcharge
if p==4 and a==1:
    peak_surcharge=150
else:
    peak_surcharge=0
pass_p_total=(net_pass_fee+p_fee+peak_surcharge)

print("--------------------------------------------------------------")
print("Module C: Tiered Electricity Utility Billing (Slab Logic)")
print("--------------------------------------------------------------")
units=float(input("Enter monthly electricity consumption(in kwh):"))

if units<0:
    print("Error units cant be negative")
    exit()
elif units<=100:
    e_bill=(units*3.00)+50
elif units<=300:
    e_bill=(units*5.00)+100
elif units<=500:
    e_bill=(units*7.50)+150
else:
    e_bill=(units*10.00)+250

#Final total
total_monthly=(pass_p_total+e_bill)

print("--------------------------------------------------------------")
print("                 CALCULATED INVOICE BREAKDOWN")
print("--------------------------------------------------------------")
print("Base Access Pass Fee :", basefee)
print("Discount             :", discount)
print("Parking Fee          :", p_fee)
print("Peak Hour Surcharge  :", peak_surcharge)
print("Net Pass & Parking Total :", net_pass_fee)
print("Electricity Units    :", units)
print("Electricity Bill     :", e_bill)
print("pass parking total     :", pass_p_total)

print("--------------------------------------------------------------")
print("TOTAL MONTHLY PAYABLE:", total_monthly)
print("--------------------------------------------------------------")
```
